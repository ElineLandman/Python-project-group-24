# Research Objective

The objective of this project is to describe how passenger traffic at Amsterdam Schiphol Airport changed during the COVID-19 pandemic and the following recovery period. We use monthly passenger statistics and reported COVID-19 cases in the Netherlands to compare long-term trends, seasonal patterns and passenger groups.

**Main research question**

*How did passenger traffic at Amsterdam Schiphol Airport change during the COVID-19 pandemic and recovery period, compared with pre-pandemic patterns?*

**Subquestions**

1. **RQ1:** How did passenger traffic and seasonal patterns at Schiphol develop between 1992 and 2019?
2. **RQ2:** How did monthly passenger traffic relate to reported COVID-19 cases in the Netherlands during 2020-2022, and how does this relationship change when using a recovery index and one- or two-month lags?
3. **RQ3:** How did recovery differ between European and intercontinental passengers, and between origin-and-destination and transfer passengers, from March 2020 to August 2026?
4. **RQ4:** How did total passenger traffic recover from March 2020 to August 2026 relative to pre-pandemic levels, and how sensitive are the results to the reference period and recovery criteria?


# 1. Introduction

The COVID-19 pandemic coincided with a major disruption in international aviation. Schiphol provides a relevant case for studying this disruption because it serves both European and intercontinental travel, including passengers who transfer between flights. Differences between these groups are relevant for understanding changes in airport demand and capacity needs.

Passenger traffic was already changing before 2020 and normally follows a seasonal pattern. A low passenger total in winter therefore has a different meaning from the same total in summer. We first examine the historical trend and seasonality, then compare passenger traffic with reported Dutch COVID-19 cases. We also compare passenger groups and assess recovery relative to pre-pandemic reference periods. Simple additional checks examine earlier case reports, yearly seasonal strength and different definitions of sustained recovery.

This is a descriptive study. The data show when and how passenger volumes changed, but cannot separate the effects of infections, travel restrictions, airline capacity, airport operations or economic conditions.

## Data pipeline and checks

The code is in `analysis_in_vscode.ipynb`. It reads the local Excel and CSV files, standardises dates and converts passenger columns to numeric values. The passenger series is checked for duplicate dates, missing months and missing or non-finite counts. All 416 months from January 1992 to August 2026 are present. The historical seasonality analysis also checks that each year in 1992-2019 contains 12 valid monthly observations.

The notebook exports only images used in the report to `figure/` and numerical summary tables to `results/`. The additional checks use the same pandas, NumPy and Matplotlib tools as the main analysis.


# 3. Methodology and Indicators

## Monthly recovery index

For calendar month *m* in year *y*, the recovery index is:

$$I_{y,m}=100\times\frac{P_{y,m}}{P_{2019,m}}$$

Here $P_{y,m}$ is the passenger total for that month. An index of 100 means the same number of passengers as in the corresponding month of 2019. For example, August 2026 is compared with August 2019. This comparison accounts for the seasonal pattern of the reference year, but is not a full statistical seasonal adjustment and does not allow for growth that might otherwise have occurred.

For passenger groups, we calculate the index separately using each group's own 2019 volume. We define sustained recovery as at least three consecutive calendar months at or above a threshold of 90% or 100%. The search starts in March 2020, excluding the months before the main disruption. Dates in the recovery table mark the **start** of the qualifying period. We also compare these results with a stricter window of six consecutive months, using the same thresholds and reference data. These are descriptive criteria, not statistical tests. A later qualifying date under a longer window does not mean a calculation error; temporary declines can interrupt the run.

To check sensitivity to the reference period, total passengers are also compared with the mean of the same calendar month in 2017-2019. Annual summaries are unweighted means of the monthly indices, not ratios of annual passenger totals. The 2026 summary includes January-August only.

## Seasonality indicator

For each year before 2020, monthly passengers are expressed relative to that year's average monthly volume:

$$S_{y,m}=100\times\frac{P_{y,m}}{(1/12)\sum_{k=1}^{12}P_{y,k}}$$

An index above 100 identifies an above-average month within that year. We average the indices for each calendar month over 1992-2019 and separately for 1992-2000, 2001-2010 and 2011-2019. The mean absolute deviation of the 12 period-level monthly indices from 100 summarises the size of the seasonal pattern. It is measured in index percentage points and does not capture all variation between individual years. Monthly totals are not adjusted for different numbers of days per month.

We also calculate seasonal strength separately for each year: the mean absolute deviation of that year's 12 indices from 100. A simple straight line fitted against year describes the direction of change over 1992-2019. This is a descriptive fit, not a significance test. Averaging yearly strength is not exactly the same as first averaging the monthly patterns and then measuring their strength; the first measure keeps more variation between years.

## COVID-19 comparison

We use the monthly arithmetic mean of available daily new confirmed cases per million residents. This is a mean daily reporting rate within a month, not a monthly case total. The analysis compares the timing of the two series and calculates Pearson correlation and descriptive linear fits for 2020-2022 together and for each year separately. The same-month yearly comparisons contain 12 observations each.

To check whether the outcome measure matters, we repeat the correlations using the 2019-relative total-passenger index instead of raw counts. We then compare passenger traffic in month *t* with reported cases in month *t*, *t-1* and *t-2*. A lag of one month therefore means earlier cases and later passenger traffic. The pooled comparisons contain 36, 35 and 34 pairs at lags zero, one and two. Within each year, pairs stay inside that calendar year, giving 12, 11 and 10 observations. Sample sizes and included months therefore differ across lags. We examine all three lags; we do not select one as proof of a delayed effect.

As a sensitivity check, the 2022 correlation is recalculated 12 times, leaving out a different month each time. This checks whether any one month determines the result. It does not remove seasonal influences, account for delayed responses or establish causality.


# 4. Results and Discussion

## 4.1 RQ1 - Long-term development and seasonality, 1992-2019

### Long-term passenger development

![Annual passenger traffic](figure/schiphol_annual_passengers.png)

*Figure 4.1: Annual passenger traffic at Schiphol, 1992-2019, in millions.*

Annual passenger traffic increased from 18.7 million in 1992 to 71.7 million in 2019, an increase of 53.0 million passengers and almost a fourfold expansion. Growth was not continuous. Traffic stayed around 39-41 million in 2000-2003 and fell from 47.7 million in 2007 to 43.5 million in 2009. After 2010, growth resumed, reaching 71.0 million in 2018 and 71.7 million in 2019.

This matters for the recovery comparison: 2019 had the highest annual volume in the historical period studied. Returning to that level does not necessarily mean returning to the growth path that might have continued without the pandemic.

### Average monthly seasonality

![Average monthly seasonality](figure/schiphol_monthly_seasonality.png)

*Figure 4.2: Mean monthly seasonal index for 1992-2019. An index of 100 represents a year's average monthly passenger volume.*

The lowest average seasonal index was 75.7 in February. The highest values were 122.1 in July and 122.2 in August. May to October were above the annual monthly average, while November to April were below it. This shows why recovery should be compared between equivalent calendar months rather than only between consecutive months.

### Changes between historical periods

![Seasonal patterns across three periods](figure/schiphol_seasonality_comparison.png)

*Figure 4.3: Mean calendar-month seasonal indices for 1992-2000, 2001-2010 and 2011-2019.*

The overall winter-low and summer-high pattern remained similar. However, the seasonal difference became somewhat smaller. The mean absolute deviation from 100 decreased from **15.35 percentage points** in 1992-2000 to **13.94** in 2001-2010 and **12.95** in 2011-2019. This is a decrease of about 15.6% between the first and last period. The underlying values are saved in `results/schiphol_seasonality_summary.csv`.

The peak index also declined from 125.0 in the first period to 120.0 in the last period. These descriptive comparisons suggest a weaker average seasonal pattern, but no statistical test was performed. RQ1 therefore shows strong long-term growth and a recurring seasonal pattern. The next check considers how its strength varies between individual years.

### Seasonal strength in individual years

![Yearly seasonal strength](figure/schiphol_annual_seasonal_strength.png)

*Figure 4.4: Mean absolute deviation of each year's monthly seasonal indices from 100, for 1992-2019. The dashed line is a descriptive linear fit against year.*

Yearly seasonal strength varies, but its fitted trend is downward by **0.142 index points per year** (r = -0.663; R² = 0.440). The means of the yearly strengths are 15.35, 14.03 and 13.05 points for the three periods. They are slightly different from the earlier period-pattern values of 15.35, 13.94 and 12.95 because the order of averaging and taking absolute deviations differs. Both descriptions show a weaker seasonal pattern in the later period, with variation between individual years. The yearly values are saved in `results/annual_seasonal_strength.csv`.

This check supports the descriptive answer to RQ1, but does not establish a statistically significant trend or explain its cause. A less pronounced seasonal pattern still leaves summer much busier than winter.


## 4.2 RQ2 - Passenger traffic and Dutch COVID-19 case reports, 2020-2022

### Timing of changes

![Passenger and case-report time series](figure/covid_netherlands_world.png)

*Figure 4.5: Monthly passengers at Schiphol (top) and monthly mean daily reported Dutch COVID-19 cases per million people (bottom), 2020-2022.*

Passenger traffic fell sharply in spring 2020 and remained low for the rest of the year. The largest reported case wave in the comparison occurred in early 2022, when passenger volumes had already recovered substantially. The timing of the largest disruption in passenger traffic therefore differed from the timing of the highest reported case rate.

### Combined and year-specific relationships

![Pooled relationship between cases and passengers](figure/covid_passengers_total.png)

*Figure 4.6: Monthly passengers and monthly mean daily reported cases per million people, pooled over 2020-2022. Each point represents one month (n = 36). Colours distinguish years; the dotted line is the pooled linear fit.*

Across all 36 months, the correlation was almost zero (r = 0.009; R² < 0.001). A single reported case level could coincide with very different passenger volumes. Combining the years also mixes the initial disruption with later recovery, so the pooled result may hide differences between years.

![Year-specific relationships between cases and passengers](figure/covid_passengers_per_year.png)

*Figure 4.7: The same relationship shown separately for 2020, 2021 and 2022 (n = 12 per year). Axes are shared to support comparison. Dotted lines show descriptive linear fits.*

In 2020 the association was weakly negative (r = -0.293; R² = 0.086). In 2021 it was close to zero (r = 0.077; R² = 0.006). In 2022 it was more strongly negative (r = -0.811; R² = 0.658). The 2022 fit accounts for about 66% of the observed variation in that year's monthly passenger totals, but this is an in-sample description rather than a causal explanation.

January-March 2022 had monthly mean daily case rates above 2,000 per million and passenger volumes of 2.5-3.7 million. To check sensitivity to single observations, we left out each 2022 month in turn. The resulting correlations ranged from **-0.856 to -0.727**. The weakest negative correlation occurred when February was omitted. The relationship therefore remained negative after removing any one month, although its strength changed. This does not rule out the combined influence of several months. The results are saved in `results/correlation_sensitivity_2022.csv`.

### Recovery index and simple time lags

![Case correlations with passenger totals and recovery indices](figure/covid_passenger_recovery_lags.png)

*Figure 4.8: Pearson correlations between earlier reported case rates and later passenger data, using raw totals (top) and the same-month 2019 recovery index (bottom). Lag 0 compares the same month, lag 1 uses cases from the preceding month and lag 2 uses cases from two months earlier. Pooled n = 36, 35 and 34; within-year n = 12, 11 and 10. All comparisons remain within 2020-2022, and year-specific pairs remain within the named year.*

Using the recovery index gives a same-month pooled correlation of **0.085**, compared with **0.009** for raw totals. Both are close to zero. In 2022, the same-month index correlation is **-0.860**, compared with **-0.811** for raw totals. Comparing equivalent calendar months therefore does not remove the difference between the pooled and year-specific results.

For the pooled recovery index, r increases to **0.215** at a one-month lag and **0.368** at a two-month lag. However, this pattern does not repeat consistently within individual years. In 2021, the index correlation changes from **0.252** at lag 0 to **-0.655** at lag 2. In 2022, it changes from **-0.860** to **-0.342**. The complete comparisons, including raw totals and sample sizes, are in `results/covid_recovery_lag_comparison.csv`.

RQ2 therefore shows no stable relationship with the same sign and strength across years, outcome measures and lags. A positive pooled lag correlation does not mean that higher case reports increased passenger traffic: recovery over time and differences between years can also contribute. Pair counts and included months change with lag. The recovery index reduces the influence of the 2019 seasonal pattern, but neither these comparisons nor the sensitivity check establish a causal effect. Conditions outside the Netherlands, travel rules and airline decisions are not tested here.


## 4.3 RQ3 - Differences in recovery between passenger groups

![Recovery by passenger category](figure/schiphol_passenger_groups.png)

*Figure 4.9: Passenger indices relative to the same calendar month in 2019, January 2020-August 2026. Top: European and intercontinental passengers. Bottom: O&D and transfer passengers. Dashed lines mark 100% of each group's own 2019 level.*

All four groups reached their lowest indices in April 2020. European and O&D traffic fell to 1.6% of their corresponding April 2019 levels, transfer traffic to 3.0% and intercontinental traffic to 3.3%.

Differences appeared during recovery. In **August 2020**, European traffic reached 33.5% of its reference level, compared with 12.3% for intercontinental traffic. In **August 2021**, the corresponding indices were 61.2% and 42.3%. The European advantage was not permanent: the lines were closer in later years, with intercontinental traffic also having the higher index in some months.

Transfer traffic recovered more strongly than O&D traffic during parts of the early recovery. In **December 2020**, the indices were 36.6% and 14.0%, respectively. Later, transfer traffic exceeded 100% in some months, but fell to 82.9% in January 2026. Such changes show why one month above a threshold is not enough to describe sustained recovery.

![Passenger-group recovery summary](figure/passenger_group_recovery_table.png)

*Table 4.1: Lowest group indices and the start of the first three consecutive months at or above 90% or 100% of the same-month 2019 reference. The search starts in March 2020 and ends in August 2026. “Not reached” means no qualifying three-month period was found.*

Transfer passengers first met the sustained 90% threshold in a period starting in **July 2023**. O&D followed in **December 2023**, and European and intercontinental passengers in **January 2024**. No group stayed at or above 100% for three consecutive months in the observed period. The full numerical table is saved in `results/passenger_group_recovery_table.csv`.

The answer to RQ3 is a severe initial decline across all groups, followed by different recovery paths. Transfer traffic reached the 90% threshold first under the three-month definition. Section 4.4 checks whether this ranking remains the same with a longer recovery window. A higher index describes stronger recovery relative to a group's own baseline, not a larger absolute passenger volume. The analysis identifies differences, but does not establish their causes.


## 4.4 RQ4 - Total passenger recovery and reference-period sensitivity

### Total recovery relative to 2019

![Passenger disruption and recovery](figure/schiphol_disruption_and_recovery.png)

*Figure 4.10: Monthly total passenger index relative to the same month in 2019, January 2020-August 2026. The dashed line marks 100%.*

The total index fell to **2.1% in April 2020**, a decline of about 97.9% compared with April 2019. Recovery was gradual and uneven. The mean monthly index was 31.7% in 2020, 34.8% in 2021 and 72.3% in 2022. These are means of monthly indices, not annual passenger-total ratios.

![Passenger recovery since 2023](figure/schiphol_recovery_since_2023.png)

*Figure 4.11: Detail of the total passenger index, January 2023-August 2026, using the same-month 2019 reference.*

The index rose from **78.0% in January 2023** to **89.7% in December 2023**, an improvement of 11.7 percentage points. Recovery continued, but there were temporary declines. In January 2026 the index fell to 87.4%. By August 2026 it had reached **97.3%**, still 2.7 percentage points below August 2019. No month from January 2023 to August 2026 reached 100% of the corresponding 2019 total.

### Sensitivity to the recovery window

![Recovery thresholds under different window lengths](figure/passenger_recovery_threshold_sensitivity.png)

*Table 4.2: Start of the first qualifying period for 90% of the same-month 2019 level over three or six consecutive months, and for 100% over six months. The search covers March 2020-August 2026. No group or total traffic reaches 100% for either three or six consecutive months. “Not reached” means no qualifying run exists in the observed period.*

Total passengers first reach the 90% threshold for three months from **January 2024**, but a six-month run starts only in **May 2024**. The European category has the same start under both windows, **January 2024**. For the six-month criterion, O&D traffic follows from **April 2024**, transfer traffic from **June 2024** and intercontinental traffic from **September 2024**.

The earlier result that transfer traffic recovered first therefore depends on the three-month definition. It does not hold with the six-month definition, where European traffic is first. In contrast, the finding that no group achieves sustained 100% recovery is unchanged under either window. The full 20 combinations of group, threshold and window length are in `results/recovery_threshold_sensitivity.csv`.

### Comparison with a broader reference period

![Recovery using two reference periods](figure/Recovery_comparison_2019_vs_2017_2019.png)

*Figure 4.12: Total passenger indices using either the corresponding month in 2019 or the average of that month in 2017-2019 as the reference.*

The two lines follow a similar pattern. The 2017-2019 reference gives slightly higher annual mean indices, with differences of 0.5-2.0 percentage points. For 2025, the mean monthly index is **96.0% using 2019** and **98.0% using 2017-2019**. For January-August 2026, the corresponding means are 95.3% and 97.3%; these should not be treated as full-year results. The numerical summaries are in `results/recovery_baseline_summary.csv`.

Both references show a large decline followed by substantial recovery. The full-year mean remains below 100% through 2025 under both comparisons. This supports the main conclusion, although the exact distance from the reference depends on the chosen baseline. Neither reference is a forecast of traffic without the pandemic.

### Monthly patterns during recovery

![Monthly passenger comparison by year](figure/schiphol_monthly_comparison.png)

*Figure 4.13: Monthly passenger totals for 2019-2025, in millions. Each line represents one year.*

In January and February 2020, volumes were similar to those in 2019, before falling sharply in March and April. In later years the winter-low and summer-high pattern became more visible again. By 2025, monthly volumes were close to the 2019 curve. This raw-count plot helps show the return of a familiar seasonal pattern; the indices above provide the more direct recovery comparison.

RQ4 therefore shows substantial but incomplete recovery relative to the selected references. Changing from 2019 to the 2017-2019 average changes the index level slightly, but not the overall finding for the completed years. Sustained recovery dates and the ranking of passenger groups do depend on whether three or six months are required, so they should always be reported with the chosen definition.


## 4.5 Discussion and limitations

Together, the results show a large interruption of Schiphol's long-term growth, followed by recovery that differed between passenger groups. The seasonal analysis shows that the disruption should be interpreted against a normal pattern of busy summers and quieter winters. For airport planning, approaching a reference total does not automatically mean that the passenger mix or demand in every month has returned to its earlier pattern. The return of busy summers suggests that capacity planning still needs a seasonal view. Differences between O&D and transfer recovery also suggest that one total-passenger indicator is not enough to describe all airport processes. These are planning considerations, not estimates of staffing or capacity needs; operational data would be needed for those decisions.

The near-zero pooled correlation and the sharp passenger decline are not contradictory. They answer different questions: the recovery index describes changes relative to a reference month, while correlation measures whether Dutch reported case rates and passenger totals moved together linearly. The relationship also changed between years. The new index and lag comparisons reinforce this point: a larger pooled correlation at a positive lag is not a consistent pattern in every year.

There are four main limitations. First, the analysis cannot isolate the causal contributions of infections, restrictions, airline decisions or other factors. Second, reported cases depend on testing and reporting practices and do not measure all infections. Dutch case reports also do not represent all countries served by Schiphol.

Third, monthly correlations combine seasonal changes and disruption or recovery over time. The 2022 sensitivity check shows that its negative association is not caused by one month alone, but it does not address these wider influences. Only one- and two-month lags were explored. Counts range from 10 to 12 observations in the within-year comparisons, and the set of included months changes with lag. Observations close in time may also be related to each other. The simple correlations and yearly seasonality fit do not model this dependence or support formal causal or significance claims.

Finally, both recovery references use historical volumes rather than a forecast of the traffic that might have occurred without the pandemic. Same-month comparisons do not remove changes in seasonality over time. The difference between three- and six-month recovery dates also shows that recovery rankings depend on an analyst-chosen definition. The latest year is incomplete. Values in the text are calculated from the local data and rounded for presentation; more digits would not remove these limitations.


# 5. Conclusion

Schiphol's annual passenger traffic increased from 18.7 million in 1992 to 71.7 million in 2019. Summer months were consistently busier than winter months. Both average monthly patterns and yearly seasonal-strength summaries show a somewhat weaker seasonal pattern in the later historical period. The yearly linear fit is descriptive and does not establish a significant trend.

In April 2020, total traffic fell to 2.1% of its April 2019 level. Recovery was uneven and differed between passenger groups. Transfer passengers reached the sustained 90% threshold first under a three-month definition, from July 2023. With six consecutive months required, European passengers were first, from January 2024. The ranking of recovery therefore depends on the definition. No group maintained 100% of its reference level for either window in the observed period.

By August 2026, total monthly traffic was at 97.3% of August 2019. The 2017-2019 reference gives slightly higher indices, but the full-year mean remained below 100% through 2025 under both references. Recovery was substantial but incomplete relative to the selected benchmarks.

The relationship with Dutch case reports changed between years, recovery measures and lags. Pooled same-month correlations were close to zero for both raw totals and the recovery index. The pooled index correlation became positive at longer lags, while year-specific patterns differed. The negative 2022 same-month correlation remained when any one month was omitted. These findings do not identify how much of the passenger change was caused by infections, restrictions or other factors.
