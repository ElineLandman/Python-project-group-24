# 1. Introduction

The COVID-19 pandemic coincided with a major disruption in international aviation. Amsterdam Schiphol Airport provides a useful case for investigating how passenger volumes changed during the pandemic and how they subsequently developed. A long historical series is important: passenger traffic was changing before 2020 and also follows a seasonal pattern, with differences between calendar months.

This project analyzes Schiphol's monthly passenger data from 1992 onwards together with reported COVID-19 cases in the Netherlands. First, it establishes the pre-pandemic trend and recurring seasonal pattern. It then examines the pandemic period, differences between passenger categories, and recovery relative to a pre-pandemic reference year.

To assess the impact of COVID-19 on passenger traffic at Schiphol, the analysis first establishes the long-term development and seasonal patterns before the pandemic. These historical patterns provide a baseline for interpreting changes during and after COVID-19. The study then examines the relationship between passenger traffic and reported COVID-19 cases, followed by differences between passenger categories. Finally, the recovery of passenger traffic is evaluated relative to pre-pandemic levels.

**Scope:** This is an exploratory, descriptive study of passenger volumes. It does not identify the causal effect of COVID-19 independently of travel restrictions, airline capacity, airport operations, or economic conditions. ## Calculation of the 2019-relative recovery index

The recovery index is used instead of absolute passenger numbers to compare passenger traffic across different years while accounting for the seasonal pattern observed in 2019. By comparing each month with the corresponding month in 2019, differences caused by the usual variation between calendar months are partially controlled for. However, this approach does not account for longer-term growth trends or changes in seasonal patterns over time.

For calendar month *m* in year *y*, the passenger index is:

$$I_{y,m}=100\times\frac{P_{y,m}}{P_{2019,m}}$$

where $P_{y,m}$ is the number of passengers in that month. An index of 100 denotes the same passenger total as the **corresponding month in 2019**. This comparison reduces the influence of ordinary month-to-month differences in the 2019 baseline, but it is **not** a full statistical seasonal adjustment and does not account for longer-term growth that would otherwise have occurred.

## Seasonality indicator (RQ1)

For each pre-pandemic year, calculate the mean monthly passenger volume and express each month relative to that annual monthly mean:

$$S_{y,m}=100\times\frac{P_{y,m}}{(1/12)\sum_{k=1}^{12}P_{y,k}}$$

Then compare the mean seasonal indices over 1992–2000, 2001–2010, and 2011–2019. A value above 100 indicates a month with above-average traffic *within that year*. The mean absolute distance of the 12 average monthly indices from 100 can summarize how pronounced the **average period-level** seasonal pattern is. It does not capture all year-to-year variability.

## COVID-19 indicator (RQ2)

Daily reported new confirmed cases per million residents are aggregated to calendar months using the arithmetic mean of available daily figures. A monthly mean is therefore **not** the total number of reported cases during that month. When interpreting the plots, note the effects of testing policy, reporting practices, and potential time lags.
